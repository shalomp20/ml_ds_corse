```python
X= [[ 148 , 24, 1385  ],
    [ 132 , 24, 2031  ],
    [ 453 , 11, 86  ],
    [ 148 , 24, 1385  ],
    [ 132 , 74, 2031  ],
    [ 453 , 31, 86  ],
    [ 148 , 24, 1385  ],
    [ 132 , 54, 2031  ],
    [ 453 , 21, 86  ]
    ]
 


X = np.array(X)
X
```

```python
ones= np.ones(X.shape[0])
ones
```

```python
np.column_stack((ones,  X))
```

```python
XTX = (X.T).dot(X)
```

```python
y = [100 , 200 , 300 , 400 , 500 , 600 , 700 , 800 , 900]
```

```python
XTX_inv = np.linalg.inv(XTX)
```

```python
XTX.dot(XTX_inv).round()
print("XTX_inv shape:", XTX_inv.shape)
print("X.T shape:", X.T.shape)
```

```python
w_full = XTX_inv.dot(X.T).dot(y)
```

```python
w0 = w_full[0]
w = w_full[1:]
```

```python
def train_linear_regression(x , y ) :
    ones = np.ones(x.shape[0])
    X = np.column_stack([ones, x])

    XTX = (X.T).dot(X)
    XTX_inv = np.linalg.inv(XTX)
    w_full = XTX_inv.dot(X.T).dot(y)
    w0 = w_full[0]
    w = w_full[1:]
    return w0 , w
```
